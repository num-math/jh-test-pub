This notebook demonstrates every standard MyST cell tag.
Tags are set via `:tags:` on a `{code-cell}` directive or in cell metadata.
Build with `myst build --execute --html` to see their full effect.

- **`remove-*`** tags strip content from the final HTML entirely.
- **`hide-*`** tags keep content in the HTML but hide it behind a clickable toggle.
- **`skip-execution`** prevents a cell from being run during `--execute`.
- **`raises-exception`** allows an erroring cell to pass without failing the build.

## 1. Baseline — no tags

A normal code cell with no tags. Both input and output appear.

```python
print("No tags — everything visible.")
```

## 2. `remove-input`

The code source is stripped from the HTML; only the output remains.

```python
print("remove-input: code hidden, output visible.")
```

## 3. `remove-output`

The output is stripped; only the code remains.

```python
print("remove-output: code visible, output hidden.")
```

## 4. `remove-cell`

The entire cell (code + output) is gone from the HTML.
Nothing from this cell will appear on the page.

```python
print("remove-cell: this whole cell is gone from the HTML.")
```

## 5. `hide-input`

The code source is hidden behind a clickable toggle.
A small "Show code" button appears; clicking it reveals the input.

```python
print("hide-input: click the toggle to reveal the code.")
```

## 6. `hide-output`

The output is hidden behind a clickable toggle.

```python
print("hide-output: click the toggle to reveal the output.")
```

## 7. `hide-cell`

The whole cell (code + output) is hidden behind a single toggle.

```python
print("hide-cell: click the toggle to reveal the entire cell.")
```

## 8. `skip-execution`

This cell is **not executed** during `myst build --execute`.
The code is shown but there will be no computed output.
This is useful for cells that prompt for user input or run expensive computations.

```python
print("skip-execution: this cell was NOT executed during the build.")
```

## 9. `raises-exception`

Normally an uncaught exception stops the build.
With this tag the cell is allowed to fail; the error message becomes the output.

```python
raise ValueError("This error is expected — the build will continue.")
```

## 10. `remove-stderr`

Standard error stream is stripped from the output.

```python
import sys
print("This goes to stdout — visible.")
print("This goes to stderr — hidden.", file=sys.stderr)
```

## 11. `remove-stdout`

Standard output stream is stripped; only stderr remains.

```python
import sys
print("This goes to stdout — hidden.")
print("This goes to stderr — visible.", file=sys.stderr)
```

## Combining tags

Multiple tags can be combined on a single cell.
Here `hide-output` is paired with `remove-stderr` to keep only clean stdout visible behind a toggle.

```python
import sys
print("Combined: output toggled, stderr removed.")
print("A noisy warning skipped.", file=sys.stderr)
```
