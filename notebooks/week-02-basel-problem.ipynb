This notebook starts from the naive series

$$
\sum_{n=1}^{\infty} \frac{1}{n^2}
$$

and compares its partial sums against the exact value $\pi^2 / 6$.

```python
import math

terms = 200
partial_sums = []
total = 0.0

for n in range(1, terms + 1):
    total += 1.0 / (n * n)
    partial_sums.append(total)

target = math.pi**2 / 6.0

print(f"S_{terms} = {partial_sums[-1]:.12f}")
print(f"pi^2 / 6 = {target:.12f}")
print(f"error    = {target - partial_sums[-1]:.12e}")
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(range(1, terms + 1), partial_sums, label=r"$\sum_{n=1}^{N} 1/n^2$")
ax.axhline(target, color="black", linestyle="--", label=r"$\pi^2 / 6$")
ax.set_xlabel("N")
ax.set_ylabel("partial sum")
ax.set_title("Naive convergence of the Basel problem")
ax.legend()
plt.show()
```

The convergence is slow but visible even for a very small, intentionally simple example.
